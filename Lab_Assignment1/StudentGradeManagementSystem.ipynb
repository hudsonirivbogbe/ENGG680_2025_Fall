```python
# Classes and Function (3 marks)

# Step 1: Create a class BankAccount.
# Add class attribute bankname="Canada Bank"
# Add 2 instance attributes: account_holder (string) & balance (float, default value 0.0)

# Step 2: Add methods
# 1. deposit(amount) -> Add amount to the balance
# 2. withdraw(amount) -> Subtract the amount if balance is sufficient, else print an error message
# 3. get_balance() -> Print Account holder's name and current balance
# 4. get_bankname() -> Print the class attribute (staticmethod)
# Note: Use f-strings to display the amount and updated balance whenever you call deposit() or withdraw()


# Step 3: Test
# 1. Print the Bankname
# 2. Create an object of BankAccount with your name and an initial balance of 1000
# 3. Deposit 500
# 4. Withdraw 200
# 5. Print the final balance


# Creating the BankAccount class
class BankAccount:
    bankname = "Canada Bank"  # Class attribute

    def __init__(self, account_holder, balance = 0.0):
        self.account_holder = account_holder
        self.balance = float(balance)

# Adding methods
    def deposit(self, amount):
        self.balance += amount
        print(f"Deposit: ${amount:.2f} | The new balance for {self.account_holder} is: ${self.balance:.2f}")

    def withdraw(self, amount):
        if amount > self.balance:
            print(f"The withdrawal failed: {self.account_holder} has insufficient funds.")
        else:
            self.balance -= amount
            print(f"Withdrawal: ${amount:.2f} | The remaining balance for {self.account_holder} is: ${self.balance:.2f}")

    def get_balance(self):
        print(f"{self.account_holder} has a balance of ${self.balance:.2f}")

    @staticmethod
    def get_bankname():
        print(f"Bank Name: {BankAccount.bankname}")

# Testing the class
BankAccount.get_bankname()          # Print the bank name
account = BankAccount("Olumide Adesanya", 1000)  # Create account with initial balance
account.deposit(500)                 # Amount of 500 Deposited
account.withdraw(200)                # Amount of 200 Withdrawn
account.get_balance()                # Output the Final balance
```

## Part 6: Linear LSE

### Observations: values of a function f(a) in range of -π to π, contaminated with random noise

### Unknown States: coefficients of polynomial functions for curve fitting

```python
# step 1
# import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# step 2
# Load data.xlsx file, which contains the observations (with noise), variance of observations and true observations
# hint: use pandas library to parse excel file and use numpy library to convert it to array
# (1 mark)
df_data =  df = pd.read_excel('data.xlsx', engine='openpyxl')
df_data.head()


arr_data = df_data.to_numpy()
print (arr_data)

#store the noisy observations, variance of observations and true observations into numpy arrays seperately
# (1 mark)
l = df_data.iloc[:, 0].to_numpy()
l_var = df_data.iloc[:, 1].to_numpy()
l_true = df_data.iloc[:, 2].to_numpy()

print(l)
print(l_var)
print(l_true)
```

```python
# step 3
# define the number of observation n
# (1 mark)
n = len(arr_data)
print(n)

# define the observation weight matrix P for LSE as identity matrix
# identity weight matrix simply means the observations will be unweighted in the following LSE
# (1 mark)
P = np.identity(n)
print (P)

# define the range of function input(i.e., constants a), from start to end; and the numerical array of a
start = -1* np.pi
end = np.pi
a = np.linspace(start, end, n)
print(a)
```

```python
# step 4
# use the matplotlib to plot the true and noisy observations together in one single figure
# use scatter for noisy observation plotting
# for the true observation, label it as "f(a)"" and set color as 'k'
# for the noisy observation scatter plot, label it as "f(a) + e", set color as 'tab:blue', set marker as 'o' and set alpha as 0.3
# include proper axis labels, legend and title
# scale the range of x-axis to be -1 to 1
# keep this format for the rest of the plots in this lab assignment
# (2 marks)

true_observation = l_true      # true function f(a)
noisy_observation = l

# Create figure and axis
fig, ax = plt.subplots(figsize=(14, 5))

# Plot of true observations
ax.plot(a/np.pi, true_observation, label='True Observations = f(a)', color='k') #("a/np.pi" indicates that the x-axis is rescaled by factor of π for better visualisation purposes)

# Scatter plot of noisy observations
ax.scatter(a/np.pi, noisy_observation, label='Noisy Observations = f(a) + e', color='tab:blue', marker='o', alpha=0.3)

# Labelling the axes and adding a title
ax.set_xlabel('Constants (a)')
ax.set_ylabel('Observations (True and Noisy)')
ax.set_title('True Observations vs Noisy Observations')

ax.legend(loc='upper left')

# Creating limits on the x-axis
ax.set_xlim(-1, 1)

# Creating a grid for the resulting plot
ax.grid(True, which='both', ls='-', color='0.65')

fig.tight_layout()
```

```python
# step 5
# define a curve fitting function named fit_polynomial_curve
# the function should implement the linear least square estimation
# the function should include all important intermediate steps in LSE
# the input and output of the funciton are described in the comment below in red.
# please keep code clean and well-commented
# (7 marks)

def fit_polynomial_curve(a: np.ndarray, l: np.ndarray, P: np.ndarray, deg: int) -> tuple:
    '''
    Args:
        a:              function input (constants)
        l:              noisy observations
        P:              observation weight matrix
        deg:            degree of polynomial to fit to data
    Returns:
        x:              estimated polynomial coefficients
        cov_x:          uncertainty in estimated coefficients
        apv:            a posterier variance factor (goodness of fit)
        v:              observation residuals
    '''
    # Step 1: This creates the design matrix for the polynomial
    A = np.vander(a, deg + 1, increasing=True)

    # Step 2: This section of the code solves the normal equations (weight matrix inclusive), which is given as (x = (AᵀPA)⁻¹ AᵀPl)
    ATA = A.T @ P @ A
    ATl = A.T @ P @ l
    x = np.linalg.solve(ATA, ATl)

    # Step 3: Compute residuals: v = Ax - l
    l_hat = A @ x
    v = l_hat - l

    # Step 4: This computes the "a posteriori variance factor", which is given as (apv = (vᵀPv) / dof)
    dof = len(l) - (deg + 1) #This computes the degree of freedom
    apv = (v.T @ P @ v) / dof

    # Step 5: Compues the estimates covariance matrix of x given as cov_x = apv * (AᵀpA)⁻¹
    cov_x = apv * np.linalg.inv(ATA)

    return (x, cov_x, apv, v)
```

```python
# step 6
# call the fit_polynomial_curve function and fetch the results
# use polynomial degrees of: 2, 4, 6
# (1 marks)

# For Degree 2
x_deg2, cov_x_deg2, apv_deg2, v_deg2  = fit_polynomial_curve(a, l, P, deg=2)
# For Degree 4
x_deg4, cov_x_deg4, apv_deg4, v_deg4 = fit_polynomial_curve(a, l, P, deg=4)
# For Degree 6
x_deg6, cov_x_deg6, apv_deg6, v_deg6 = fit_polynomial_curve(a, l, P, deg=6)

print("Coefficients for degree 2:", x_deg2)
print("Coefficients for degree 4:", x_deg4)
print("Coefficients for degree 6:", x_deg6)
```

```python
# step 7
# use the estimated coefficients x and constants array a, construct polynomial functions for each degree
# hint: use polyval function from numpy library (check input argument carefully)
# (1 marks)

quadratic_uw = np.polyval(x_deg2[::-1], a)
quartic_uw   = np.polyval(x_deg4[::-1], a)
hexic_uw     = np.polyval(x_deg6[::-1], a)
```
